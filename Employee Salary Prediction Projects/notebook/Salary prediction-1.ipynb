### Definig x,y Variable

Create A Feature x And Target y In The Dataframe

```python
x=data[['Years','Job Rate']]
y=data['Monthly Salary']
print(x,y)
```

### train_test_split Is Imported From sklearn.model_selection Using The train_test_split

split into a test 20% and training 80%

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
print(x_train)
print(x_test)
```

### LinearRegessionm Is Imported From Sklearn.linear_model Using The LinearRegression

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

### Making  LinearRegression prediction

```python
predlr=lr.predict(x_test)
predlr
```

### Actual Vs Prediction

```python
print("Actual:", y_test.values[:10])
print("predicted:", predlr[:10])
```

###  Visualize The Regression Line
#### plot The Actual Data point
#### Plot The Regression  Line
#### Add Labels And Title

```python
plt.scatter(x_test['Years'], y_test, color='blue',label='Actual')
plt.scatter(x_test['Years'], predlr, color="red", label='Regression Line')
plt.title("Linear Regression Years Salary vs Experience")
plt.xlabel('Experience')
plt.ylabel('Annual Salary')
plt.legend()
plt.show()
```

#### mean_absolute_error,mean_squared_error is importred Fom sklearn.metrix Using The mean_absolute_error,mean_squared_error

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,root_mean_squared_error
```

Evaluatio The Model To Calculate using MAE,MSE

```python
print("MAE:", mean_absolute_error(y_test, predlr))
print("MSE:", mean_squared_error(y_test, predlr))
print("RMSE:", root_mean_squared_error(y_test, predlr))
print("R2:", r2_score(y_test,predlr))
```
